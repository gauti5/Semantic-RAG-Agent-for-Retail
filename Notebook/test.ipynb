```python

from langchain_community.document_loaders import UnstructuredPDFLoader

class PDFLoader:
    def __init__(self, pdf_path: str):
        self.pdf_path=pdf_path
        
    def load_pdf(self):
        print("="*50)
        print("Loading PDF...")
        print("="*50)
        
        loader=UnstructuredPDFLoader(self.pdf_path)
        documents=loader.load()
        print(f"Total pages loaded : {len(documents)}")
        
        return documents
    def metadata(self, documents):
        print("\nAdding metadata..")
        for i, doc in enumerate(documents):
            doc.metadat['page']=i+1
            doc.metadat['document name']=self.pdf_path.split('/')[-1]
            
        print("Metadata Addedd Successfully!")
        
    def preview(self, documents):
        print("\nPreview : ")
        print("="*50)
        
        for doc in documents[:3]:
            print(f"\nPage : {doc.metadata['page']}")
            print("-"*50)
            
            print(doc.page_content[:300])
            print('\n')
    
    def load(self):
        documents=self.load_pdf()
        documents=self.metadata(documents=documents)
        self.preview(documents=documents)
        
        return documents
        
```

```python
PDF_PATH='C:/Users/ssandeep011/Downloads/Semantic RAG Agent for Retail/Data/Retail_Display_Manual.pdf'

loader=PDFLoader(pdf_path=PDF_PATH)
documents=loader.load()
print()
print(f"Total pages loaded : {len(documents)}")
```
